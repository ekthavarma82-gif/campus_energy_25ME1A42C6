```
#program3: a function to calculate electricity cost
def calculate_bill(kwh,rate_per_unit):
    return kwh * rate_per_unit
todays_bill = calculate_bill(500,8)
print("Today's estimated bill is Rs.", todays_bill)
```

```
import pandas as pd

# Load the dataset
df = pd.read_csv('campus_energy.csv')

# Remove extra spaces from column names
df.columns = df.columns.str.strip()

# Display the column names
print("Column names:", df.columns.tolist())

# Display the first few rows
print("\nHere is the raw data:")
display(df.head())

# Calculate total usage per BUILDING
building_totals = df.groupby('BUILDING')['KWH_USE_PER_Hour_usuage'].sum()

print("\nTotal usage per Building:")
print(building_totals)

# Generate a statistical summary
print("\nStatistical Summary:")
display(df.describe())
```
